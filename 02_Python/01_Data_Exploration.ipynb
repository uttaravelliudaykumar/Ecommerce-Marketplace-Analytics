```python
#  Customer Repeat Purchase Analysis

# 1. Count total customer records
total_customer_records = customers["customer_id"].nunique()

# 2. Count actual unique customers
total_unique_customers = customers["customer_unique_id"].nunique()

print("Total Customer Records:", total_customer_records)
print("Total Unique Customers:", total_unique_customers)


# 3. Connect customer information with order information
customer_orders = orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

# 4. Count orders placed by each unique customer
orders_per_customer = (
    customer_orders
    .groupby("customer_unique_id")["order_id"]
    .nunique()
)

# 5. Identify repeat customers
repeat_customers = orders_per_customer[
    orders_per_customer > 1
]

# 6. Count one-time customers
one_time_customers = orders_per_customer[
    orders_per_customer == 1
]

# 7. Calculate repeat customer percentage
repeat_customer_percentage = (
    len(repeat_customers) / len(orders_per_customer)
) * 100

# 8. Display results
print("\nCustomer Purchase Analysis")
print("--------------------------")
print("Total Unique Customers:", len(orders_per_customer))
print("One-Time Customers:", len(one_time_customers))
print("Repeat Customers:", len(repeat_customers))
print(
    "Repeat Customer Percentage:",
    round(repeat_customer_percentage, 2), "%"
)

# 9. Find customers with the highest number of orders
top_repeat_customers = (
    orders_per_customer
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Customers by Number of Orders:")
print(top_repeat_customers)
```

```python
# Step 19: Customer Order Frequency Analysis

# 1. Count orders for every unique customer

orders_per_customer = (
    customer_orders
    .groupby("customer_unique_id")["order_id"]
    .nunique()
)

# 2. Count how many customers placed 1, 2, 3, etc. orders

order_frequency = (
    orders_per_customer
    .value_counts()
    .sort_index()
)

print("Customer Order Frequency:")
print(order_frequency)


# 3. Calculate average orders per customer

average_orders = orders_per_customer.mean()

print("\nAverage Orders Per Customer:", round(average_orders, 2))


# 4. Find maximum orders placed by one customer

maximum_orders = orders_per_customer.max()

print("Maximum Orders by One Customer:", maximum_orders)


# 5. Calculate percentage of customers by order frequency

order_frequency_percentage = (
    order_frequency / len(orders_per_customer)
) * 100

print("\nCustomer Order Frequency Percentage:")
print(order_frequency_percentage.round(2))
```

```python
# Step 20: Customer Review and Satisfaction Analysis

# 1. Count reviews by score

review_distribution = reviews["review_score"].value_counts().sort_index()

print("Review Score Distribution:")
print(review_distribution)


# 2. Calculate percentage of each review score

review_percentage = (
    reviews["review_score"].value_counts(normalize=True).sort_index() * 100
)

print("\nReview Score Percentage:")
print(review_percentage.round(2))


# 3. Calculate average review score

average_review_score = reviews["review_score"].mean()

print("\nAverage Review Score:", round(average_review_score, 2))


# 4. Categorize reviews into positive, neutral and negative

def categorize_review(score):
    if score >= 4:
        return "Positive"
    elif score == 3:
        return "Neutral"
    else:
        return "Negative"


reviews["review_category"] = reviews["review_score"].apply(categorize_review)


# 5. Count positive, neutral and negative reviews

review_categories = reviews["review_category"].value_counts()

print("\nReview Categories:")
print(review_categories)


# 6. Calculate percentage by review category

review_category_percentage = (
    reviews["review_category"].value_counts(normalize=True) * 100
)

print("\nReview Category Percentage:")
print(review_category_percentage.round(2))


# 7. Count reviews with and without written comments

reviews_with_comments = reviews["review_comment_message"].notna().sum()

reviews_without_comments = reviews["review_comment_message"].isna().sum()

print("\nReviews With Written Comments:", reviews_with_comments)
print("Reviews Without Written Comments:", reviews_without_comments)
```

```python
# Step 21: Delivery Performance Analysis

# 1. Create a separate copy of the orders dataset
delivery_analysis = orders.copy()

# 2. Convert date columns into proper datetime format
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    delivery_analysis[column] = pd.to_datetime(
        delivery_analysis[column],
        errors="coerce"
    )

# 3. Select only delivered orders with available dates
delivered_orders = delivery_analysis[
    (delivery_analysis["order_status"] == "delivered") &
    (delivery_analysis["order_delivered_customer_date"].notna()) &
    (delivery_analysis["order_estimated_delivery_date"].notna()) &
    (delivery_analysis["order_purchase_timestamp"].notna())
].copy()

# 4. Calculate actual delivery time in days
delivered_orders["actual_delivery_days"] = (
    delivered_orders["order_delivered_customer_date"] -
    delivered_orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# 5. Calculate delivery difference against estimated date
# Positive = Late, Negative = Early, Zero = On time

delivered_orders["delivery_delay_days"] = (
    delivered_orders["order_delivered_customer_date"] -
    delivered_orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# 6. Categorize delivery performance

delivered_orders["delivery_performance"] = np.where(
    delivered_orders["delivery_delay_days"] > 0,
    "Late",
    "On Time or Early"
)

# 7. Calculate overall delivery metrics

total_delivered_orders = len(delivered_orders)

average_delivery_days = delivered_orders["actual_delivery_days"].mean()

average_delay_days = delivered_orders["delivery_delay_days"].mean()

print("DELIVERY PERFORMANCE SUMMARY")
print("-----------------------------")
print("Delivered Orders Analyzed:", total_delivered_orders)
print("Average Actual Delivery Time:", round(average_delivery_days, 2), "days")
print("Average Difference from Estimated Date:", round(average_delay_days, 2), "days")

# 8. Count late and on-time/early orders

delivery_counts = delivered_orders["delivery_performance"].value_counts()

print("\nDelivery Performance Distribution:")
print(delivery_counts)

# 9. Calculate delivery performance percentages

delivery_percentages = (
    delivered_orders["delivery_performance"].value_counts(normalize=True) * 100
)

print("\nDelivery Performance Percentages:")
print(delivery_percentages.round(2))

# 10. Find average delay among late orders only

late_orders = delivered_orders[
    delivered_orders["delivery_delay_days"] > 0
]

if len(late_orders) > 0:
    average_late_days = late_orders["delivery_delay_days"].mean()
else:
    average_late_days = 0

print("\nLate Orders:", len(late_orders))
print("Average Days Late:", round(average_late_days, 2))

# 11. Display the 10 most delayed orders

most_delayed_orders = (
    delivered_orders[
        delivered_orders["delivery_delay_days"] > 0
    ][["order_id", "order_delivered_customer_date",
       "order_estimated_delivery_date", "delivery_delay_days"]]
    .sort_values("delivery_delay_days", ascending=False)
    .head(10)
)

print("\n10 Most Delayed Orders:")
print(most_delayed_orders)
```

```python
# Step 22: Seller Performance Analysis

# 1. Combine order items with seller information

seller_data = order_items.merge(
    sellers,
    on="seller_id",
    how="left"
)

print("Seller Data Shape:", seller_data.shape)


# 2. Calculate seller performance

seller_performance = (
    seller_data
    .groupby("seller_id")
    .agg(
        total_items_sold=("order_item_id", "count"),
        total_orders=("order_id", "nunique"),
        total_revenue=("price", "sum"),
        total_freight=("freight_value", "sum"),
        seller_city=("seller_city", "first"),
        seller_state=("seller_state", "first")
    )
    .reset_index()
)

print("\nSeller Performance Sample:")
print(seller_performance.head())


# 3. Find the top 10 sellers by revenue

top_sellers_revenue = (
    seller_performance
    .sort_values("total_revenue", ascending=False)
    .head(10)
)

print("\nTop 10 Sellers by Revenue:")
print(
    top_sellers_revenue[
        ["seller_id", "seller_city", "seller_state",
         "total_orders", "total_items_sold", "total_revenue"]
    ]
)


# 4. Find the top 10 sellers by number of orders

top_sellers_orders = (
    seller_performance
    .sort_values("total_orders", ascending=False)
    .head(10)
)

print("\nTop 10 Sellers by Number of Orders:")
print(
    top_sellers_orders[
        ["seller_id", "seller_city", "seller_state",
         "total_orders", "total_revenue"]
    ]
)


# 5. Count registered sellers by state

seller_state_counts = (
    sellers["seller_state"]
    .value_counts()
)

print("\nRegistered Sellers by State:")
print(seller_state_counts)


# 6. Calculate total seller revenue

total_seller_revenue = seller_performance["total_revenue"].sum()

print("\nTotal Seller Product Revenue:", round(total_seller_revenue, 2))


# 7. Calculate average revenue per seller

average_seller_revenue = seller_performance["total_revenue"].mean()

print("Average Revenue Per Seller:", round(average_seller_revenue, 2))
```

```python
# Step 23: Product Performance Analysis

# 1. Combine order items with product information

product_sales = order_items.merge(
    products,
    on="product_id",
    how="left"
)

# 2. Add English product category names

product_sales = product_sales.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

print("Product Sales Dataset Shape:", product_sales.shape)


# 3. Calculate performance for each product

product_performance = (
    product_sales
    .groupby("product_id")
    .agg(
        total_items_sold=("order_item_id", "count"),
        total_orders=("order_id", "nunique"),
        total_revenue=("price", "sum"),
        average_price=("price", "mean"),
        product_category=("product_category_name_english", "first")
    )
    .reset_index()
)

print("\nProduct Performance Sample:")
print(product_performance.head())


# 4. Find the top 10 products by revenue

top_products_revenue = (
    product_performance
    .sort_values("total_revenue", ascending=False)
    .head(10)
)

print("\nTop 10 Products by Revenue:")
print(top_products_revenue)


# 5. Find the top 10 products by quantity sold

top_products_quantity = (
    product_performance
    .sort_values("total_items_sold", ascending=False)
    .head(10)
)

print("\nTop 10 Products by Quantity Sold:")
print(top_products_quantity)


# 6. Find the top 10 categories by number of items sold

category_quantity = (
    product_sales
    .groupby("product_category_name_english")["order_item_id"]
    .count()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Categories by Items Sold:")
print(category_quantity)


# 7. Find the average product price

average_product_price = product_sales["price"].mean()

print("\nAverage Product Price:", round(average_product_price, 2))


# 8. Count unique products sold

unique_products_sold = product_sales["product_id"].nunique()

print("Unique Products Sold:", unique_products_sold)
```

```python
# Step 24: Sales Trend Analysis

# 1. Create a copy of orders
sales_trend = orders.copy()

# 2. Convert purchase timestamp into datetime
sales_trend["order_purchase_timestamp"] = pd.to_datetime(
    sales_trend["order_purchase_timestamp"],
    errors="coerce"
)

# 3. Keep delivered orders for this sales analysis
sales_trend = sales_trend[
    sales_trend["order_status"] == "delivered"
].copy()

# 4. Create year and month columns
sales_trend["year"] = sales_trend[
    "order_purchase_timestamp"
].dt.year

sales_trend["month"] = sales_trend[
    "order_purchase_timestamp"
].dt.month

sales_trend["year_month"] = sales_trend[
    "order_purchase_timestamp"
].dt.to_period("M")

# 5. Connect orders with product sales
monthly_sales_data = sales_trend.merge(
    order_items[["order_id", "price"]],
    on="order_id",
    how="left"
)

# 6. Calculate monthly revenue
monthly_revenue = (
    monthly_sales_data
    .groupby("year_month")["price"]
    .sum()
)

print("Monthly Revenue:")
print(monthly_revenue.round(2))

# 7. Calculate monthly order count
monthly_orders = (
    sales_trend
    .groupby("year_month")["order_id"]
    .nunique()
)

print("\nMonthly Order Count:")
print(monthly_orders)

# 8. Calculate yearly revenue
yearly_revenue = (
    monthly_sales_data
    .groupby("year")["price"]
    .sum()
)

print("\nYearly Revenue:")
print(yearly_revenue.round(2))

# 9. Identify the highest revenue month
highest_revenue_month = monthly_revenue.idxmax()
highest_month_revenue = monthly_revenue.max()

print("\nHighest Revenue Month:", highest_revenue_month)
print("Highest Monthly Revenue:", round(highest_month_revenue, 2))

# 10. Identify the highest revenue year
highest_revenue_year = yearly_revenue.idxmax()

print("\nHighest Revenue Year:", highest_revenue_year)
```

```python
# Step 25: Prepare Working Copies for Data Cleaning

# 1. Create copies of all original datasets

customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
payments_clean = payments.copy()
reviews_clean = reviews.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
geolocation_clean = geolocation.copy()
category_translation_clean = category_translation.copy()

print("Working copies created successfully!")


# 2. Identify date columns in the Orders dataset

order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

# 3. Convert order date columns into datetime format

for column in order_date_columns:
    orders_clean[column] = pd.to_datetime(
        orders_clean[column],
        errors="coerce"
    )

print("\nOrders date columns converted successfully!")


# 4. Convert date columns in Order Items

order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"],
    errors="coerce"
)


# 5. Convert date columns in Reviews

review_date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for column in review_date_columns:
    reviews_clean[column] = pd.to_datetime(
        reviews_clean[column],
        errors="coerce"
    )

print("Order Items and Reviews date columns converted successfully!")


# 6. Check the new data types

print("\nOrders Data Types:")
print(orders_clean[order_date_columns].dtypes)

print("\nOrder Items Data Type:")
print(order_items_clean["shipping_limit_date"].dtype)

print("\nReviews Data Types:")
print(reviews_clean[review_date_columns].dtypes)


# 7. Verify that original data remains unchanged

print("\nOriginal Orders Data Types:")
print(orders[order_date_columns].dtypes)
```

```python
# Step 26: Missing Value Analysis

# 1. Store all cleaned datasets in a dictionary
clean_datasets = {
    "Customers": customers_clean,
    "Orders": orders_clean,
    "Order Items": order_items_clean,
    "Payments": payments_clean,
    "Reviews": reviews_clean,
    "Products": products_clean,
    "Sellers": sellers_clean,
    "Geolocation": geolocation_clean,
    "Category Translation": category_translation_clean
}

# 2. Check missing values in every dataset
for name, df in clean_datasets.items():

    print("\n" + "=" * 60)
    print(f"MISSING VALUE ANALYSIS: {name}")
    print("=" * 60)

    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing Values": missing_count,
        "Missing Percentage": missing_percentage.round(2)
    })

    missing_summary = missing_summary[
        missing_summary["Missing Values"] > 0
    ]

    if missing_summary.empty:
        print("No missing values found.")
    else:
        print(missing_summary.sort_values(
            by="Missing Values",
            ascending=False
        ))

# 3. Check missing values in Orders by order status
print("\n" + "=" * 60)
print("MISSING DELIVERY DATES BY ORDER STATUS")
print("=" * 60)

delivery_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

print(
    orders_clean.groupby("order_status")[delivery_columns]
    .apply(lambda group: group.isnull().sum())
)

# 4. Check missing product categories
print("\n" + "=" * 60)
print("PRODUCT CATEGORY MISSING VALUE CHECK")
print("=" * 60)

print(
    "Missing product categories:",
    products_clean["product_category_name"].isnull().sum()
)

# 5. Check missing review comments
print("\n" + "=" * 60)
print("REVIEW COMMENT MISSING VALUE CHECK")
print("=" * 60)

print(
    "Missing review titles:",
    reviews_clean["review_comment_title"].isnull().sum()
)

print(
    "Missing review messages:",
    reviews_clean["review_comment_message"].isnull().sum()
)
```
